# Analyse des ventes avec Python

## 1. Présentation du projet

L'objectif de ce projet est d'analyser les ventes de trois produits : un ordinateur, un téléphone et un casque.

Cette analyse permet de calculer le chiffre d'affaires total, d'étudier les quantités vendues et de comparer la contribution de chaque produit aux ventes totales.

## 2. Outils utilisés

- **Python** : langage de programmation utilisé pour l'analyse.
- **Pandas** : bibliothèque utilisée pour organiser et manipuler les données.

## 3. Préparation des données

Nous allons créer un tableau contenant le nom des produits, leur prix et les quantités vendues. Ensuite, nous calculerons le chiffre d'affaires de chaque produit.

```python
import pandas as pd

data = {
    "Produit": ["Ordinateur", "Téléphone", "Casque"],
    "Prix": [1200, 800, 100],
    "Quantite": [2, 3, 5]
}

df = pd.DataFrame(data)
df["Chiffre_affaires"] = df["Prix"] * df["Quantite"]

print(df)
print("Chiffre d'affaires total :", df["Chiffre_affaires"].sum())
```

```python
import pandas as pd

data = {
    "Produit": ["Ordinateur", "Téléphone", "Casque"],
    "Prix": [1200, 800, 100],
    "Quantite": [2, 3, 5]
}

df = pd.DataFrame(data)

df["Chiffre_affaires"] = df["Prix"] * df["Quantite"]

print(df)

print("Chiffre d'affaires total :", df["Chiffre_affaires"].sum())
```

## 4. Statistiques descriptives

Cette étape permet de résumer les données numériques à l'aide de statistiques telles que la moyenne, le minimum et le maximum.

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "produit": ["ordinateur", "téléphone", "casque"],
    "prix": [1200, 800, 100],
    "quantite": [2, 3, 5]
}

df = pd.DataFrame(data)
df["chiffre_affaires"] = df["prix"] * df["quantite"]
print(df)
print("\nStatistiques descriptives:")
print(df.describe())
```

## 5. Analyse des produits

Nous allons identifier le produit qui génère le chiffre d'affaires le plus élevé et celui qui est vendu en plus grande quantité.


```python

ligne_max = df.loc[df["chiffre_affaires"].idxmax()]

print("\nLe produit avec le chiffre d'affaires le plus élevé est :", ligne_max["produit"])
print("Chiffre d'affaires :", ligne_max["chiffre_affaires"])
produit_plus_vendu = df.loc[df["quantite"].idxmax()]

print("produit le plus vendu :", produit_plus_vendu["produit"])
print("quantite vendue :", produit_plus_vendu["quantite"])
```

## 6. Visualisation des ventes

Les graphiques permettent de comparer visuellement le chiffre d'affaires et les quantités vendues pour chaque produit.

```python

plt.figure(figsize=(8, 5))

plt.bar(df["produit"], df["chiffre_affaires"])

plt.title("chiffre d'affaires par produit")
plt.xlabel("produit")
plt.ylabel("chiffre d'affaires")

plt.show()
```

## 7. Conclusion

Le chiffre d'affaires total des trois produits est de 5 300. L'ordinateur et le téléphone génèrent chacun un chiffre d'affaires de 2 400, tandis que le casque génère 500.

Le casque est le produit le plus vendu en quantité, avec 5 unités. Cependant, il contribue moins au chiffre d'affaires total.
